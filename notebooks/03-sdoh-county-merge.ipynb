# SDOH county merge

Scaffold notebook generated for this repository.

```
import pandas as pd
patients = pd.DataFrame({"patient_id":[1,2,3], "county_fips":["06037","36061","48201"], "outcome":[0,1,1]})
sdoh = pd.DataFrame({"county_fips":["06037","36061","48201"], "svi_percentile":[0.82,0.65,0.74], "unemployment_rate":[5.1,4.2,4.9]})
merged = patients.merge(sdoh, on="county_fips", how="left")
merged
```

```
assert merged["svi_percentile"].notna().all()
print("Ready for feature derivation")
```

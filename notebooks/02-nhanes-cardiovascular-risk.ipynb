```
import pandas as pd

# Replace with NHANES component merges keyed by SEQN and correct survey weights.
df = pd.DataFrame({
    "age": [35, 52, 61, 44, 70],
    "systolic_bp": [118, 142, 155, 130, 160],
    "bmi": [24.5, 31.0, 29.0, 26.2, 33.1],
    "high_risk": [0, 1, 1, 0, 1],
})
df
```

```
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(max_depth=2, random_state=42).fit(df[["age","systolic_bp","bmi"]], df["high_risk"])
model.score(df[["age","systolic_bp","bmi"]], df["high_risk"])
```
